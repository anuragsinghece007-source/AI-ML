# Daily practise problem module challenge 1.

**First Iteration: num = 10Condition: 10 % 10 == 0 \(\rightarrow \) True (10 is perfectly divisible by 10)Operation: num // 2 \(\rightarrow \) 10 // 2 = 5Update Tuple: result += (5,) \(\rightarrow \) result becomes (5,)**

```python
numbers = [10, 15, 20, 25]
result = ()

for num in numbers:
  if num % 10 == 0:
    result += (num // 2,)

print(result)
print(sum(result))
```

```python
s = "VIDHISINGH"
new = " "
for i in s:
  if ord(i) >= 65 and ord(i) <= 90:
    new += chr(ord(i) + 32)
  else:
    new += i
print(new)
```
