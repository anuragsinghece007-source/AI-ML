```python
def largest(a, b):
  if a > b:
    return a
  else:
    return b
print(largest(12, 25))
```

```python
def total(numbers):
  s = 0
  for n in numbers:
    s+=n

  return s

print(total([1, 2, 3, 4, 5]))
```
