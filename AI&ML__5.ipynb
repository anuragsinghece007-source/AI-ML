```python
#  Body Mass Index (BMI) is a simple measure that helps determine whether a
# person's weight is appropriate for their height.
# Write a Python program that takes the BMI value as input from the user
# and displays the corresponding weight category using the following rules:
# BMI Value Category
# Less than 18.5 Underweight
# 18.5 to 24.9 Normal Weight
# 25.0 to 29.9 Overweight
# 30.0 or above Obese


bmi = float(input("Enter your BMI value: "))

if bmi < 18.5:
  print("underweight")
elif bmi <= 24.9:
  print("normal wieght")
elif bmi <= 29.9:
  print("overwieght")
else:
  print("obese")

```

```python
#  Body Mass Index (BMI) is a simple measure that helps determine whether a
# person's weight is appropriate for their height.
# Write a Python program that takes the BMI value as input from the user
# and displays the corresponding weight category using the following rules:
# BMI Value Category
# Less than 18.5 Underweight
# 18.5 to 24.9 Normal Weight
# 25.0 to 29.9 Overweight
# 30.0 or above Obese
bmi = float(input("Enter your BMI value: "))
if bmi < 18.5:
    print("Underweight")
elif bmi >= 18.5 and bmi <= 24.9:
    print("Normal Weight")
elif bmi >= 25.0 and bmi <= 29.9:
    print("Overweight")
else:
    print("Obese")
```
