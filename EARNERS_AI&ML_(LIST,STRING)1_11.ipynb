```python
numbers = [1, 2, 3, 4] # Initial State   numbers = [1, 2, 3, 4]

n = len(numbers)#  n = len(numbers) → n = 4
for i in range(n): # range(n) yani range(4) generate karega: 0, 1, 2, 3
  numbers[i] = numbers[i] + i # Calculation: 1 + 0 = 1


print(numbers)
```

```python
marks = [65, 75, 85]
total = 0
for m in marks: # m = 65 , 75 , 85
  if m >= 70:
    total += m

print(total)

# Loop ExecutionIteration
# m = 65
# Condition Check: if 65 >= 70:
# Action: The if block is skipped. total remains 0.
```

```python
names = ["Amit", "Neha", "Aditya"]
longest = names[0]
for name in names:
  if len(name) > len(longest):
    longest = name
print(longest)
```

```python
data = [10, 20, 30]
for i in range(len(data)):
  data[i] = data[i] + 10
print(max(data))
print(min(data))
```

```python
numbers = [5, 10, 15, 20]
total = 0
for i in range(len(numbers)):
    if i % 2 == 0:
      total += numbers[i]
print(total)
```
